```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
docs = [
    "LangChain is a framework for developing LLM applications.",
    "RAG fetches relevant external context to minimize hallucinations.",
    "Fine-tuning adapts model weights on domain-specific datasets.",
]
```

```python
# 2. Embeddings & Vector Store (Replaces all NumPy math)
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
vectorstore = InMemoryVectorStore.from_texts(docs, embedding=embeddings)
retriever = vectorstore.as_retriever(search_kwargs={"k": 2})
```

```python
# 3.Prompt Template (Clean & Reusable)
prompt = ChatPromptTemplate.from_messages([
    (
        "system", 
        "Answer the user's question based only on the provided context."
    ),
    (
        "user", 
        "Context: {context}\n\nQuestion: {question}"
    )
])
```

```python
# 4. LLM
llm = ChatOpenAI(model="gpt-4o-mini")
```

```python
chain= prompt | llm | StrOutputParser()
```

```python
# 6. Execute
user_query = "What helps reduce LLM hallucinations?"
docs=retriever.invoke(user_query)
```

```python
docs
```

```python
retrival_page_content="/n/n".join([ doc.page_content for doc in docs ])
```

```python

response = chain.invoke({"context":retrival_page_content, "question": user_query})

print(response)
```
